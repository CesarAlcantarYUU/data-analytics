## Lab 1. Working with names - parsing strings

```python
test_name = 'John Doe'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name [:i]
            last = name[i+1:]
            print(f"Full Name:  {name}")
            print(f"First Name: {first}")
            print(f"Last Name:  {last}")
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0])
```

```python
parse_and_display(test_list[1])
```

```python
parse_and_display(test_list[2])
```

```python
parse_and_display('Cesar Alcantar')
```

```python
def parse_and_display2(name):
    names = name.split()
    print(f"Full Name: {name}")

    if len(names) == 1:
        print(f"First Name: {names[0]}")
    elif len(names) == 2:
        print(f"First Name: {names[0]}")
        print(f"Last Name: {names[1]}")

    elif len(names) == 3:
        print(f"First Name: {names[0]}")
        print(f"Middle Name: {names[1]}")
        print(f"Last Name: {names[2]}")
```

```python
test2_list = ['lorde', 'Billie Eilish', 'Megan thee Stallion']
for name in test2_list:
    parse_and_display2(name)
    print()
```
