This lab focuses on defining custom Python functions using parameters and return statements.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician/band is {music}."
```

```python
print(favorite_things("Cesar", "The Wolf of Wall Street", "The Weeknd"))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here("I joined Year Up United to develop my data analytics skills and build a career where I can combine my marketing experience with data-driven decision making."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Hawaii", "I enjoy the beaches and beautiful scenery"))
```

```python
print(favorite_place("San Francisco", "I enjoy the city, restaurants, and different things to explore"))
```
