```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))
print(r_2)
```

```python
r_3 = np.random.rand(2, 1)
print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))
print(r_4)
```

```python
r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array([
    "Arash", "Abddul", "Faiz", "Timothy",
    "Earl", "Isreal", "Maream", "Stephanie",
    "Netsanet", "Meric", "Joshua", "Matthew"
])
print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4), replace=False)
print(breakout_groups)

#This could be an effective way to choose breakout rooms because it randomizes it to make it feel fair and equal.
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    replace=True,
    p=[0.00, 0.10, 0.10, 0.10, 0.10, 0.10,
        0.00, 0.10, 0.10, 0.10, 0.10, 0.10,]
)

print(call_on_weighted)
```

## Generating a Normal Distribution

```python
ball_curve = np.random.normal(
    loc=10,
    scale=5,
    size=(4, 5)
)
print(ball_curve)
```
