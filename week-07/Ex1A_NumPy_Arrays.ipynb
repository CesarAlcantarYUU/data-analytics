```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13],
                 [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

# The First loop prints arrays and the second prints individual numbers

```python
for x in arrA:
    for y in x:
        loop_result = x * 2

    print(loop_result)
```

### x represents an entire row and the result is printed once for each row, not once for each element.

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.vstack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
stacked_h = np.hstack((arrA, arrA))
# hstack joins the arrays horizontally, adding columns.
print(stacked_h)

stacked_v = np.vstack((arrA, arrA))
# vstack joins the arrays vertically, adding rows.
print(stacked_v)

stacked_d = np.dstack((arrA, arrA))
# dstack joins the arrays along the third dimension.
print(stacked_d)
```

## Splitting Arrays

```python
splitsky = np.split(joined_array_1, 2)

print(splitsky)

# This creates 2 arrays.
# Each array is 2-dimensional and has 2 rows and 3 columns.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# This creates 3 arrays.
# Each array is 2-dimensional.
# Because there are 4 rows, the arrays contain 2, 1, and 1 rows.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, "odd")

print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6],
                 [13, 5],
                 [7, 3]])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
print(arrB)

y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
